```python
import threading
import time
import random

# 1. DEFINIÇÃO DE THREADS (Processos Leves)
# Imagine que este é um processo de "Editor de Texto"
def tarefa_digitacao():
    print("[Thread Usuário] ESTADO: Em Execução. Aguardando sua entrada...")
    # O input() coloca a thread no ESTADO: Bloqueado/Espera
    nome = input("Digite seu nome para o documento: ")
    print(f"\n[Thread Usuário] Documento atualizado para: {nome}")

def tarefa_auto_save():
    # Simula o Pseudoparalelismo: Rodando 'ao mesmo tempo' que a digitação
    for i in range(3):
        # ESTADO: Pronto -> Em Execução
        print(f"\n[Thread Sistema] ESTADO: Em Execução. Fazendo Auto-Save {i+1}...")
        time.sleep(4) # Simula o tempo que o SO leva para outras tarefas
    # ESTADO: Finalizado
    print("\n[Thread Sistema] Tarefa de Auto-Save concluída.")

# 2. IMPLEMENTAÇÃO (O SO criando os fluxos dentro do processo)
print("--- INICIANDO PROCESSO DO EDITOR ---")

# Criamos as threads que compartilham os mesmos recursos
thread_1 = threading.Thread(target=tarefa_digitacao)
thread_2 = threading.Thread(target=tarefa_auto_save)

# 3. CICLO DE VIDA (Início)
thread_1.start()
thread_2.start()

# 4. TABELA DE PROCESSOS (Sincronização)
# O comando .join() garante que o SO espere as threads terminarem para fechar o processo pai
thread_1.join()
thread_2.join()

# 5. CICLO DE VIDA (Finalização)
print("\n--- PROCESSO FINALIZADO PELO SISTEMA OPERACIONAL ---")
```
